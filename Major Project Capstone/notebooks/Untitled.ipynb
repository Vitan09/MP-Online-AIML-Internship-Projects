```python
import os
import shutil
import kagglehub

# Create local dataset folder
project_dataset = "../dataset"

os.makedirs(project_dataset, exist_ok=True)

print("Downloading Fashion-MNIST...")

# Download from KaggleHub
download_path = kagglehub.dataset_download(
    "zalando-research/fashionmnist"
)

print("Downloaded to:", download_path)

# Copy files into project dataset folder
for file in os.listdir(download_path):

    source = os.path.join(download_path, file)
    destination = os.path.join(project_dataset, file)

    if not os.path.exists(destination):
        shutil.copy2(source, destination)

print("\nDataset copied to:")
print(os.path.abspath(project_dataset))
```

```python
import os
import numpy as np

dataset_path = "../dataset"

train_images = os.path.join(dataset_path, "train-images-idx3-ubyte")
train_labels = os.path.join(dataset_path, "train-labels-idx1-ubyte")

test_images = os.path.join(dataset_path, "t10k-images-idx3-ubyte")
test_labels = os.path.join(dataset_path, "t10k-labels-idx1-ubyte")
```

```python
def load_images(filepath):

    with open(filepath, "rb") as file:

        images = np.frombuffer(
            file.read(),
            dtype=np.uint8,
            offset=16
        )

    images = images.reshape(-1, 28, 28)

    return images


def load_labels(filepath):

    with open(filepath, "rb") as file:

        labels = np.frombuffer(
            file.read(),
            dtype=np.uint8,
            offset=8
        )

    return labels
```

```python
x_train = load_images(train_images)
y_train = load_labels(train_labels)

x_test = load_images(test_images)
y_test = load_labels(test_labels)

print("Training Images :", x_train.shape)
print("Training Labels :", y_train.shape)

print("Testing Images :", x_test.shape)
print("Testing Labels :", y_test.shape)
```
